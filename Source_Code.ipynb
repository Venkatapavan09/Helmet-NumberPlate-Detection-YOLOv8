```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

results = model.train(
    data="helmet_plate/data.yaml",
    epochs=10,
    imgsz=640,
    batch=4,
    project="runs",
    name="helmet_plate_project",
    device="cpu",
    workers=0,
    cache=False,
    verbose=True
)
```

```python
metrics = model.val(
    data = "helmet_plate/data.yaml",
    device="cpu",
workers = 0,
batch = 4,
verbose=True)


print('map50-95 :',metrics.box.map)
print('map50 :',metrics.box.map50)
print('precision :',metrics.box.mp)
print('recall :',metrics.box.mr)
```
